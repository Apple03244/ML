## Baysian Decision Theory

먼저, $Y = \{c_1,c_2,...,c_n\}$이 있고, $\lambda_{ij}$는 실제 레이블이 $c_j$에서 $c_i$로 잘못 분류했을 경우

발생하는 손실을 나타냅니다

아래의 수식은 조건 리스크를 수식으로 나타낸 것입니다.

$R(c_i|x)=\sum_{j=1}^{n}\lambda_{ij}P(c_j|x)$

결국 우리의 목적은 조건리스크의 합(평균)을 최소로 만드는 기준 $h$를 찾는 것입니다.

$R(h)= \mathbb{E_x}[R(h(x)|x)) ]$

결국 아래의 $h^{*}$를 베이즈 최적 분류기라고 부르고 이애 대응하는 전체 리스크 $R(h^*)$ 를 베이즈 리스크라고 부릅니다

$h^*(x)= \argmin_{c \in Y} R(c|x)$

### 사후확률

베이지안 결정 이론을 사용하기 위해선 사후확률 $P(c|x)$를 알아야함을 알수있습니다.

우리는 아래 두 가지 전력을 고려할 수 있습니다.(직접 사후 확률을 얻는 것은 현실적으로 매우 어렵습니다)

1. 판별모델
    - 직접 $P(c|x)$f를 모델링 하는 방법

2. 생성모델
    - 결합 확률 분포 $P(x,c)$를 먼저 모델링한 후 이를 통해 $P(c|x)$를 얻는 방법

우리가 앞서 살펴본 많은 방법은 판별 모델에 속하고 이번엔 생성 모델의 방법은 자세히 살펴보겠습니다.

### 생성모델

당연한 얘기지만 조건부 확률이 전제 조건이 되어야합니다.

$P(c|x)=\frac{P(c,x)}{P(x)}$

여기에 베이즈 정리를 추가하면 다음과 같이 표현할 수 있습니다.

$P(c|x)=\frac{P(x|c)P(c)}{P(x)}$

위의 수식에서 보이는 각각의 확률을 정리하면

1. $P(x)$ 는 증거팩터로써 클래스 $c$와는 무관합니다.
2. $P(c)$ 는 사전확률
3. $P(x|c)$는 클래스 조건 확률(우도)라고 합니다.

이 때 우도 $P(x|c)$ 는 $x$에 대한 모든 속성의 결합확률을 알아야하는데 이는 현실적으로 어렵습니다.

- 속성 $x$가 $d$개의 이진속성들로만 이루어져있다고 해도 결합확률은 $2^d$입니다

### 최대 우도 추정
우도를 구하기 위해 자주 사용되는 전략은 먼저 모종의 정해진 확률 분포 형식이 있다고 가정하고,
훈련 샘플에 기반해 확률 분포의 파라미터를 추정하는 전략입니다.

파라미터 벡터 $\theta_c$ 에 의해서만 결졍된다고 가정하여 우리는 파라미터 벡터 $\theta_c$ 를 추정하게 됩니다.

가장 전통적인 방법인 *최대 우도 추정법*을 소개하겠습니다

#### Maximum Likelihood Estimation

$D_c$는 훈련세트 $D$에서 클래스 $c$에 해당해는 샘플들의 집합이고, 이들이 독립항등분포라고 가정한다면,

$P(D_c|\theta_c) = \prod_{x \in D_C} P(x|\theta_c)$

위의 수식을 만족합니다

곱셈은 오버플로우 가능성떄문에 일반적으로 로그우도(Log Likelihood)를 사용합니다

$LL(\theta_c) = \log{P(D_c|\theta_c)} = \Sigma_{x \in D_c}{\log{P(x|\theta_c)}}$

이 때 최대우도 $\hat\theta_c =  \argmax_{\theta_c}{LL(\theta_c)}$ 입니다.

### Naive Bayes Classifier

베이즈 공식을 이용한 사후확률 $P(c|x)$을 계산하기 위해 $P(x|c)$를 사용하고 이는 모든 속성의 결합확률을 사용하는 것이기 때문에

유한한 샘플 공간에서 이를 추정하기 쉽지않습니다.

이런 문제를 해결하고자 Naive Bayes Classifier 는 속성 조건독립 가정(attrubute conditional independence assumption)을 활용합니다.

수식으로 살펴보면 이해가 쉽습니다.

$P(c|x)=\frac{P(c)P(x|c)}{P(x)}=\frac{P(c)}{P(x)} \prod_{x_i \in D}{P(x_i|c)}$

모든 클래스에 대해 $P(x)$는 동일하므로 아래와 같이 표현할 수 있습니다.

$h_{nb}(x) = \argmax_{c \in Y}{P(c)\prod_{x_i \in D}{P(x_i|c)}}$

이게 Naive Bayes Classifier 의 표현식입니다.

표현식을 좀 더 자세히 알아봅시다.

클래스 c의 부분집합, 즉 샘플공간 $D_c$에 대해서 아래와 같이 나타낼 수 있습니다.

$P(c)=\frac{|D_c|}{|D|}$

만일 $x_i$가 이산속성이면,

$P(x_i|c)=\frac{|D_{c,x_i}|}{|D_c|}$
