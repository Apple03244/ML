## Margin

샘플 공간에서 분할 초평면은 다음 선형 방정식을 통해 묘사될 수 있습니다.

$w^T x+b=0$

임의의 점 $x'$에 대해 분할 초평면까지의 거리는 다음과 같습니다.

$r=\frac{|w^T x' +b|}{||w'||}$

위의 초평면이 훈련 샘플을 완벽하게 분류할 수 있다면 다음과 같습니다.

만약 샘플 $y_i \in \{-1,1\}$를 정학하게 분류할 수 있는 초평면이 있다면 다음과 같습니다.

$w^Tx_i+b>= +1 , y_i= +1$

$w^Tx_j+b<= -1 , y_j=-1$

이 떄, 초평면에 가장 가까운 몇개의 훈련 샘플을 *서포트 벡터* 라고 합니다.

두 개의 서로 다른 클래스의 서포트 벡터에서 초평면까지의 거리 합을 *마진(margin)* 이라고 합니다.
