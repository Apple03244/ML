## 연속값

굉장히 쉬운 아이디어로 간단하게 구현할 수 있습니다.

연속 속성 a가 샘플내에서 가진 값을 오름차순, 혹은 내림차순으로 정렬합니다.

$a=\{a_1,a_2,a_3,...,a_n\}$

그렇다면 아래와 같은 기준들을 만들 수 있습니다.

$T_a=\{\frac{a_t+a_{t+1}}{2}|t=1,2,3...,n-1\}$

## 결측값

두가지 조건을 고민해야합니다.

1. 결측값이 있는 속성의 분할 속성을 어떻게 정할 것인가?
2. 분할 속성을 정한경우 샘플의 해당 속성값이 결측값일 경우, 샘플을 어떻게 분할할 것인가?

우선 훈련세트 $D$, 속성 $a$를 가정하고, $\bar D$를 속성 $a$의 무결측 부분집합이라고 표기하겠습니다.

만약 속성 $a$가 다음과 같이 $\{a_1,a_2,...,a_v\}$의 속성값을 택할 수 있다면, 다음과 같이 부분집합들을 정의할 수 있습니다.

1. $\bar D^v$ 는 $\bar D$에서 속성 $a$가 $a_v$인 부분집합입니다.

2. $\bar D_k$는 속성 $k$($k=1,2,..,|y|$)인 부분집합입니다.

따라서 우리는 다음들을 정리할 수 있습니다.($w_x$는 각 샘플 $w$에 가해지는 가중치입니다)

1. $p=\frac{\sum_{x \in \bar D} w_x}{\sum_{x in D} w_x}$

2. $\bar p_k=\frac{\sum_{x \in \bar D_k} w_x}{\sum_{x in \bar D} w_x}$

3. $\bar r_v=\frac{\sum_{x \in \bar D^v} w_x}{\sum_{x in \bar D} w_x}$

우리는 정보이득을 다음과 같이 계산해야합니다.

$Gain(D,a)=p*Gain(\bar D,a)=p*(Ent(\bar D)-\sum_{t=1}^{v} \bar r_v Ent(\bar D^v))$

또한,

$Ent(\bar D)=-\sum \bar p_k \log_{2}{\bar p_k}$

으로 계산됩니다
